```python
import pandas as pd
import matplotlib.pyplot as ply
import seaborn as sns
import re

positive_words = ["ভালো","সুন্দর","চমৎকার","খুশি","আনন্দ","দারুণ","অসাধারণ","ভালোবাসা","সেরা","মজার","উত্তম","শ্রেষ্ঠ","প্রিয়","সফল","আশা",]

negative_words = [
    "খারাপ",
    "দুঃখ",
    "ভয়",
    "ঘৃণা",
    "বিরক্ত",
    "সমস্যা",
    "ব্যর্থ",
    "কষ্ট",
    "রাগ",
    "অসুস্থ",
    "মন্দ",
    "নিরাশ",
    "ক্ষতি",
    "দুর্বল",
    "ভুল",]

def preprocess_text(text):
    """
    Clean Bengali text before analysis.
    - Removes numbers and punctuation
    - Strips extra whitespace
    """

    text = re.sub(r'[a-zA-Z0-9]', '', text)


    text = re.sub(r'[।,!?."\'()\-]', '', text)


    text = text.strip()

    return text

def analyse_sentiment(text):
    """
    Analyse sentiment of a Bengali sentence.
    Returns: sentiment label, score, matched words
    """
    cleaned = preprocess_text(text)
    words = cleaned.split()

    score = 0
    matched = []

    for word in words:
        if word in positive_words:
            score += 1
            matched.append((word, "Positive"))
        elif word in negative_words:
            score -= 1
            matched.append((word, "Negative"))


    if score > 0:
        label = "Positive"
    elif score < 0:
        label = "Negative"
    else:
        label = "Neutral"

    return label, score, matched

sample_sentences = [
    "এই বইটি সত্যিই দারুণ এবং অসাধারণ।",
    "আজকের দিনটি খুবই খারাপ ছিল, অনেক কষ্ট হয়েছে।",
    "আমি বাজারে গেলাম।",
    "তোমার সাথে থাকতে পেরে আমি খুশি এবং আনন্দিত।",
    "এই কাজে আমি সম্পূর্ণ ব্যর্থ এবং নিরাশ।",
    "সুন্দর আবহাওয়া, ভালোবাসা ও আশার দিন।",
    "সে খুব রাগী এবং বিরক্তিকর।",
    "আজকে কিছু মনে নেই।", ]


results = []

print("=" * 60)
print("      BENGALI SENTIMENT ANALYSIS RESULTS")
print("=" * 60)

for sentence in sample_sentences:
    label, score, matched = analyse_sentiment(sentence)

    print(f"\nSentence : {sentence}")
    print(f"Cleaned  : {preprocess_text(sentence)}")
    print(f"Matched  : {matched if matched else 'None'}")
    print(f"Score    : {score}")
    print(f"Sentiment: {label}")

    results.append({
        "Sentence": sentence[:40] + "..." if len(sentence) > 40 else sentence,
        "Score": score,
        "Sentiment": label
    })

df = pd.DataFrame(results)
```
